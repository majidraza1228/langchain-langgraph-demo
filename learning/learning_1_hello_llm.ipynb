## Step 1: Setup

First, let's import the required libraries and set up our API key.

```python
# Load environment variables (API keys)
from dotenv import load_dotenv
load_dotenv()

# Import the ChatOpenAI class
from langchain_openai import ChatOpenAI

print("Setup complete!")
```

## Step 2: Create an LLM Instance

The `ChatOpenAI` class is our connection to OpenAI's models.

```python
# Create an LLM instance
llm = ChatOpenAI(
    model="gpt-4o-mini",  # The model to use
    temperature=0.7       # Creativity level (0=focused, 1=creative)
)

print(f"LLM created: {llm.model_name}")
```

## Step 3: Send a Simple Message

Use `.invoke()` to send a message and get a response.

```python
# Send a simple string message
response = llm.invoke("What is Python in one sentence?")

print("Response type:", type(response))
print("Content:", response.content)
```

## Step 4: Using Message Objects

LangChain uses message objects to represent different types of messages:
- `SystemMessage`: Instructions for the AI
- `HumanMessage`: User's input
- `AIMessage`: AI's response

```python
from langchain_core.messages import SystemMessage, HumanMessage

# Create a conversation with system and human messages
messages = [
    SystemMessage(content="You are a helpful assistant that speaks like a pirate."),
    HumanMessage(content="What's the weather like today?")
]

response = llm.invoke(messages)
print(response.content)
```

## Step 5: Streaming Responses

For long responses, you can stream the output token by token.

```python
# Stream the response
print("Streaming response:")
for chunk in llm.stream("Tell me a short joke about programming."):
    print(chunk.content, end="", flush=True)
print()  # New line at the end
```
