````python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o-mini")
print("Setup complete!")
````

## Why Prompt Templates?

Instead of writing prompts like this every time:
```python
f"Translate '{text}' to {language}"
```

Templates let you:
- Reuse the same structure with different inputs
- Validate that all required variables are provided
- Compose complex prompts from smaller pieces

## Simple String Template

````python
# Create a simple template
template = PromptTemplate.from_template(
    "Translate the following text to {language}: {text}"
)

# See what variables it expects
print("Input variables:", template.input_variables)

# Format the template
formatted = template.format(language="French", text="Hello, how are you?")
print("Formatted:", formatted)
````

````python
# Use it with the LLM
prompt = template.format(language="Spanish", text="Good morning!")
response = llm.invoke(prompt)
print(response.content)
````

## ChatPromptTemplate (Recommended)

For chat models, use `ChatPromptTemplate` which supports system/human/ai messages.

````python
# Create a chat prompt template
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role} assistant."),
    ("human", "{question}")
])

print("Input variables:", chat_template.input_variables)
````

````python
# Format and see the messages
messages = chat_template.format_messages(
    role="cooking",
    question="How do I make pasta?"
)

for msg in messages:
    print(f"{msg.type}: {msg.content}")
````

````python
# Use with LLM
response = llm.invoke(messages)
print(response.content)
````

## Quick Template Creation

For simple cases, use `from_template` shortcut.

````python
# Quick single-message template
quick_template = ChatPromptTemplate.from_template(
    "Explain {concept} in simple terms for a {audience}."
)

messages = quick_template.format_messages(
    concept="machine learning",
    audience="5-year-old"
)

response = llm.invoke(messages)
print(response.content)
````

## Partial Templates

Pre-fill some variables to create specialized templates.

````python
# Base template
base_template = ChatPromptTemplate.from_messages([
    ("system", "You are a {language} language tutor."),
    ("human", "How do I say '{phrase}' in {language}?")
])

# Create a Spanish-specific template
spanish_template = base_template.partial(language="Spanish")

# Now only need to provide 'phrase'
messages = spanish_template.format_messages(phrase="Thank you very much")
response = llm.invoke(messages)
print(response.content)
````

## Template with Examples (Few-Shot)

Include examples to guide the LLM's response format.

````python
few_shot_template = ChatPromptTemplate.from_messages([
    ("system", "You convert text to emojis. Follow the examples."),
    ("human", "I love pizza"),
    ("ai", "I ❤️ 🍕"),
    ("human", "The weather is sunny"),
    ("ai", "The weather is ☀️"),
    ("human", "{input}")
])

messages = few_shot_template.format_messages(input="I'm going to the beach")
response = llm.invoke(messages)
print(response.content)
````
